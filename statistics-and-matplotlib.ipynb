# **Statistics**

```python
# Finding the mean
scores = [23, 45, 65, 76, 78, 65, 67, 78, 98, 67, 56, 56, 45,90]
mean_score = stats.mean(scores)
print('Mean Scores')
print(mean_score)
```

```python
# Finding the median
data = [ 23, 20, 40, 50, 60, 70]
median = stats.median(data)
print('Median of the dataset is:')
print(median)
```

```python
# Finding the mode
data = [20, 30, 40, 20]
mode = stats.mode(data)
print('The mode of the dataset is:')
print(mode)
```

```python
# Finding the variance and standard deviation
dog_heights_mm = {
    'Labrador Retriver' : 570,
'German Shepherd' : 600,
'Golden Retriver' : 700,
'Bulldog' : 400,
'Beagle': 360,
'Poodle' : 200
}
# Extract the heights into a list
heights_mm = list(dog_heights_mm.values())
# Calculating variance and standard deviation
variance_mm = stats.variance(heights_mm)
st_dev = stats.stdev(heights_mm)
# Displaying the results
print("Results:")
print(variance_mm)
print(st_dev)
```

## **Line Chart**

```python
x = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']
y = [12, 34, 56, 34, 78, 98, 67, 56, 79, 34, 54, 78]
plt.plot(x, y, label = 'Car Sales', color = 'Green', linestyle = '-')
# Adding x-labels and y-labels and title
plt.xlabel('Months')
plt.ylabel('Number of Cars Sold')
plt.title('Car Sales Data')
plt.grid(True, linestyle = '--')
plt.xticks(rotation = 45)
plt.legend()
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y1 = [10, 20, 30, 40, 50]
y2 = [15, 25, 35, 45, 55]
y3 = [20, 30, 40, 50, 60]
# Plotting the graphs
plt.plot(x, y1, marker = 'o', linestyle = '-', color = 'blue', label = 'Solid line with circles')
plt.plot(x, y2, marker = 's', linestyle = '--', color = 'Red', label = 'Dashed red line')
plt.plot(x, y3, marker = '^', linestyle = ':', color = 'Green', label = 'Green dotted line')
plt.title('Graphs comparision')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.legend()
plt.show()
```

## **Scatter Plot**

```python
# 1. Properly create the DataFrame and save it to CSV
data = {
    'Months': ['January', 'February', 'March', 'April', 'May', 'June', 
               'July', 'August', 'September', 'October', 'November', 'December'],
    'Rainfall(mm)': [12, 34, 56, 67, 45, 67, 56, 87, 98, 67, 67, 98]
}
df = pd.DataFrame(data)
df.to_csv('rainfall.csv', index=False)

# 2. Read the CSV and verify
df = pd.read_csv('rainfall.csv')
print(df.head())

# 3. Plot the histogram
plt.figure(figsize=(8, 5))
plt.hist(df['Rainfall(mm)'], bins=6, color='Green', edgecolor='Green')
plt.title('Rainfall Distributions')
plt.xlabel('Rainfall (mm)')
plt.grid(True)
plt.ylabel('Frequency')
plt.show()
```

## **Creating Subplots**

```python
# sample data
subjects = ['Maths', 'Science', 'English', 'Artificial Intelligence']
scores = [34, 56, 76, 89]
attendance = [90, 78, 67, 95]
# Create different plots
plt.figure(figsize = (12,8))
# Bar Plot
plt.subplot(2,2,1)
plt.bar(subjects, scores)
plt.title('Subject Scores')
# Line Plot
plt.subplot(2,2,2)
plt.plot(subjects, attendance, marker = 'o', color = 'green', linestyle = '--')
plt.title('Subject Attendance')
# Scatter Plot
plt.subplot(2,2,3)
plt.scatter(scores, attendance)
plt.title('Scores vs Attendance')
# Bar Plot
plt.subplot(2,2,4)
plt.pie(scores, labels = subjects, autopct = '%1.1f%%')
plt.title('Scores Distributions')
plt.tight_layout()
plt.show()
```
