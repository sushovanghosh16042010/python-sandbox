````python
print("--- Dynamic Monthly Grocery Bill ---")

grocery_items = []

while True:
    name = input("Enter item name (or type 'done' to finish): ")
    if name.lower() == 'done':
        break

    try:
        price = float(input(f"Enter price for {name}: "))
        grocery_items.append({"name": name, "price": price})
    except ValueError:
        print("Invalid price. Please enter a number.")

tax_rate = float(input("Enter Tax percentage (%): "))

# Calculations
subtotal = sum(item['price'] for item in grocery_items)
tax_amount = subtotal * (tax_rate / 100)
total_bill = subtotal + tax_amount

# Display Table
print("\n{:<20} {:>10}".format("Item", "Price"))
print("-" * 31)
for item in grocery_items:
    print("{:<20} {:>10.2f}".format(item['name'], item['price']))

print("-" * 31)
print("{:<20} {:>10.2f}".format("Subtotal", subtotal))
print("{:<20} {:>10.2f}".format(f"Tax ({tax_rate}%)", tax_amount))
print("=" * 31)
print("{:<20} {:>10.2f}".format("TOTAL BILL", total_bill))
````

**Problem**

Ask the user for their total monthly budget.
Then, in a loop, ask them to input various expenses (e.g., 'Rent: 1200', 'Groceries: 300', 'Entertainment: 100'). They should be able to enter as many expenses as they want, and type 'done' when finished.
After all expenses are entered, calculate the total amount spent.
Finally, tell the user their total expenses and how much money they have left from their budget. Also, let them know if they are over budget or under budget.

````python
print("Monthly Budget Report")
print("---------------------")
Total = int(input("Enter your total monthly budget: "))
spent = 0 # Variable to track total expenses

print("Your monthly budget is:", Total)
print("----------------------")

while True:
  print("\nOption selector")
  print("1. New entry")
  print("2. Exit and Discard")
  print("3. Done and Generate Report")

  x = int(input("Enter the option number: "))

  if x == 1:
    category = input("Enter the name of the category: ")
    price = int(input("Enter the amount spent on that category: "))
    spent += price
    print(f"Added {category}: {price}. Current total spent: {spent}")

  elif x == 3:
    print("\n--- Final Budget Report ---")
    print(f"Total Budget: {Total}")
    print(f"Total Spent:  {spent}")
    remaining = Total - spent

    if remaining >= 0:
      print(f"You are under budget! You have {remaining} left.")
    else:
      print(f"You are over budget by {abs(remaining)}!")
    break

  else:
    print("Shutting down... entries not saved.")
    break
````

````python
import numpy as np
# understanding Missing Values
# Creating a pandas DataFrame
df = pd.DataFrame({'Companies Name' : ['Google', 'Microsoft', 'Samsung'], 'Origin Country' : ['USA', np.nan, 'South Korea']})
print(df)
print("-------------------------------------------------------------------------------------------------------------------")
# Identifying missing values
print(df.isnull())
# Dropping rows with missing values
print("DataFrame after dropping rows with missing values: ")
print(df.dropna())
print("-------------------------------------------------------------------------------------------------------------------")
# Again creating that dataframe
# Creating a pandas DataFrame
df = pd.DataFrame({'Companies Name' : ['Google', 'Microsoft', 'Samsung'], 'Origin Country' : ['USA', np.nan, 'South Korea']})
print(df)
print("-------------------------------------------------------------------------------------------------------------------")
# Now filling it with missing values
# Hint: Remember that .copy is a method that needs to be 'called' with parentheses!
df_filled = df.copy()
df_filled['Origin Country'] = df_filled['Origin Country'].fillna('USA')
print("After filling the missing values: ")
print(df_filled)
print("-------------------------------------------------------------------------------------------------------------------")
````

````python
# Control Flow with loop
# Basic loop with range
print("Numbers from 1 to 5:")
for i in range(1,6):
   print(i)

# Loop through a list
fruits = ['Apple', 'Banana', 'Cherry']
print("Fruits")
for fruit in fruits:
  print(fruit)

# Loop with enumerate
print("Fruit Enumeration")
for index, fruit in enumerate(fruits):
  print(f"{index}:{fruit}")
````

## **Experiment - 5**

````python
# CSV File Operations
# read and write data to CSV files
import csv
# write data to csv
data = [['Name','Age','City'],
        ['Alice',25,'New York'],
        ['Bob',30,'London']]
with open('data.csv', 'w', newline='') as file:
  writer = csv.writer(file)
  writer.writerows(data)

# read and print CSV
print("Reading the CSV file: ")
with open("data.csv", "r") as file:
  reader = csv.reader(file)
  for row in reader:
    print(row)
````

## **Experiment - 6**

````python
# handling Missing values in a DataFrame
import pandas as pd
import numpy as np

# Create DataFrame with missing values
data = {'A' : [1, 2, np.nan, 4], 'B' : [5, np.nan, 7, 8], 'C' : [9, 10, 11, 12]}
df = pd.DataFrame(data)

# handle missing values
df_filled = df.fillna(0)
df_dropped = df.dropna()

print("Original dataFrame: ")
print(df)
print("Filled DataFrame: ")
print(df_filled)
print("Dropped DataFrame: ")
print(df_dropped)
````

## **Experiment - 7**

````python
# using list comprehension for data transformation
# create a list of numbers
numbers = [1, 2, 3, 4, 5]

# Square numbers using list comprehension
squares = [x**2 for x in numbers]

# create a list of tuples
pairs = [(x, x**2) for x in numbers]

# Filter even numbers
evens = [x for x in numbers if x % 2 == 0]

print("Original numbers: ", numbers)
print("Squares: ", squares)
print("Even Numbers: ", evens)
print("Number-square pairs: ", pairs)
````
