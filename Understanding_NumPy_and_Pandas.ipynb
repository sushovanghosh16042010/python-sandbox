## **Array**

Array is a 'data type'. An array is used to store multiple values using single identifier.

```python
# Creating an array from list
array1 = np.array([10, 20, 30])
print(array1)
# This was also an 1 dimensional arry
array2 = np.array([12, 34, 'a', 7.2])
print(array2)
# All vales have been promoted to string
```

## **Loading Arrays From Files**

Sometimes, we may have data in files and we may need to load that data in an array for processing. numpy. loadtxt () and numpy. genfromtxt () are the two functions that can be used to load data from text files. The most commonly used file type to handle large amount of data is called CSV (Comma Separated Values).

Each row in the text file must have the same number of values in order to load data from a text file into a numpy array. Let us say we have the following data in a text file named 'data.csv'

```python
file_load = np.loadtxt('data.csv', skiprows = 1, delimiter = ',', dtype = int)
print(file_load) # This will load the file
```

Some common terms
*   'skiprows' = 1 indicates that the first row is the header that is being skipped.
*   'Delimiter' species wheather the values are seperated by comma, tab or semicolon.
*   'dtype' represents the data type.

### **Using NumPy.genfromtxt()**

```python
data_array = np.genfromtxt('data_new.csv', skip_header = 1, delimiter = ',')
print(data_array)
```

# **Pandas Library**

##**Series**

```python
# creating a Pandas series
data = pd.Series([10, 20, 30], index = ['a', 'b', 'c'])
print("Pandas Series")
print(data)
```

```python
# Creating a Pandas dataFrame
# Creating a DataFrame from a dictionary
# Here we are creating a DataFrame showing various Operating Systems and their developers
# We will also keep some data empty, so we will import Numpy
import numpy as np
df = pd.DataFrame({'Operating Systems' : ['Windows', 'Android', 'MacOS', 'iOS & iPadOS', 'Linux distros'], 'Developer': ['Microsoft', 'Google', 'Apple', 'Apple', np.nan]})
print(df)
```

```python
# Adding a new row
new_row = {'Operating Systems' : 'ChromeOS', 'Developer' : 'Google'}
df = pd.concat([df, pd.DataFrame([new_row])], ignore_index = True)
print(df)
```

```python
# Deleting "MacOS" row
df = df.drop(index = 2)
print("After deleating a row: ")
print(df)
```

```python
# accessing data using loc and iloc
print(df.loc[0, 'Operating Systems'])
print(df.iloc[1,1])
```

```python
# Understanding Missing Values
# Here we will find missing values in 'df'
print(df.isnull())
```

```python
# Filling Missing Values
df_filled = df.copy()
# Fill missing "Developer" with "Open Source"
df_filled['Developer'] = df_filled['Developer'].fillna('Open Source')
print("After filling data:")
print(df_filled)
```

```python
# Creating Pandas series
series1 = pd.Series([2, 3, 4], index = ['Feb', 'Mar', 'Apr'])
print(series1)
```

```python
# Creation of Series from NumPy Arrays
array1 = np.array([1, 2, 3, 4])
series2 = pd.Series(array1)
print(series2)
print("------------------")
# We can also use letters ar strings as indices
series3 = pd.Series(array1, index = ['Jan', 'Feb', 'Mar', 'Apr'])
print(series3)
```

```python
# Creation of Series from Dictionary
dict1 = {'India' : 'New Delhi', 'UK' : 'London', 'US' : 'Washington DC'}
series4 = pd.Series(dict1)
print(series4)
```

##**DataFrame**

###**Creation of a DataFrame**

```python
# Creation of a DataFrame from NumPy ndarrays
array1 = np.array([10, 20, 30])
array2 = np.array([100, 200, 300])
array3 = np.array([-10, -20, -30, -40])
df1 = pd.DataFrame(array1)
print(df1)
print("------------------")
df2 = pd.DataFrame([array1, array3, array2], columns = ['A', 'B', 'C', 'D'])
print(df2)
```

```python
# Creation of a DataFrame from dictionary of series
resultsheet = {'Arnab' : pd.Series([90, 91, 97], index = ['Maths', 'Science', 'AI']), 'Ramit' : pd.Series([23, 45, 99], index = ['Maths', 'Science', 'AI']), 'XYZ' : pd.Series([45, 67, 99], index = ['Maths', 'Science', 'AI'])}
df9 = pd.DataFrame(resultsheet)
print(df9)
```

### **Operations on Rows and Columns in DatafRame**



```python
# Adding a new column
df9['ABC'] = [67, 89, 99]
print(df9)
```

```python
# Updating existing column
df9['Ramit'] = [99, 99, 99]
print(df9)
```

```python
# Adding a new row
df9.loc['English'] = [45, 67, 89, 90]
print(df9)
```

### **Joining, Merging and Concatenation of DataFrame**

#### **Joining**
We can use the pandas. DataFrame.append() method to merge two DataFrames. It appends rows of the second DataFrame at the end of the first DataFrame. Columns not present in the first DataFrame are added as new columns. For example, consider the two DataFrames- df10 and df11 described below. Let us use the append() method.

```python
df10 = pd.DataFrame([[1, 2, 3], [4, 5], [6]], columns = ['C1', 'C2', 'C3'], index = ['R1', 'R2', 'R3'])
print(df10)
print("-----------------")
df11 = pd.DataFrame([[10, 20], [30], [40, 50]], columns = ['C2', 'C5'], index = ['R4', 'R2', 'R5'])
print(df11)
print("-----------------")
# .append() is depreciated in Pandas 2.0+
```
