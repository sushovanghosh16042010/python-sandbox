```python
# Run this cell to setup the notebook
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.datasets import load_iris
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.cluster import KMeans
print('Notebook setup succedeed')
```

# **Pearson Correlation Coefficient**

```python
# Create a dataset
data = {'Study_Hours': [1, 2, 3, 4, 5, 6], 'Exam_Scores' : [35, 40, 50, 55, 60, 75]}
# Create a DataFrame
df = pd.DataFrame(data)
print(df)
# Calculate Pearson's r
r = df['Study_Hours'].corr(df['Exam_Scores'])
print('Pearson r')
print(r)
```

# **Supervised Learning**

## **Linear Regression**

```python
# Marks vs Hours of Study
# Create dataset
data = {
 'Study_Hours': [1, 2, 3, 4, 5, 6],
 'Marks': [30, 40, 50, 60, 70, 80]
}
# Create DataFrame
df = pd.DataFrame(data)
print(df)
# Separate input and output
X = df[['Study_Hours']]
y = df['Marks']
# Create model
model = LinearRegression()
# Train the model
model.fit(X, y)
# Predict marks for 7 hours of study
predicted_marks = model.predict([[7]])

print('Predicted Marks:')
print(predicted_marks[0])
```

```python
# Area vs House Price
# Create dataset
data = {
 'Area': [500, 700, 900, 1100, 1300, 1500],
 'Price': [15, 21, 27, 33, 39, 45]
}
# Create DataFrame
df = pd.DataFrame(data)
print(df)
# Input and output
X = df[['Area']]
y = df['Price']
# Create Linear Regression model
model = LinearRegression()
# Train the model
model.fit(X, y)
# Predict price for a 1000 sq. ft. house
predicted_price = model.predict([[1000]])
print("Predicted house price:", predicted_price[0], "Lakhs")
```

## **Classification**
In the following example we will be using 'Iris dataset', which have been imported at the starting of this notebook...

```python
# Load Iris dataset
iris = load_iris()

# Create DataFrame
df = pd.DataFrame(iris.data, columns = iris.feature_names)
df['Species'] = iris.target

print(df.head())

# Separate input and output
x = df[iris.feature_names]
y = df['Species']

# Divide data into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)

# Create the KNN Model
model = KNeighborsClassifier(n_neighbors = 5)

# Train the model
model.fit(x_train, y_train)

# Predict Species
y_pred = model.predict(x_test)

# Calculate accuracy score
accuracy = accuracy_score(y_test, y_pred)

print('Accuracy:', accuracy * 100, '%')

# Predict new flower
new_flower = pd.DataFrame([[5.1, 3.5, 1.4, 0.2]], columns = iris.feature_names)
prediction = model.predict(new_flower)

print('Predicted flower species:')
print(iris.target_names[prediction[0]])
```

# **Unsupervised Learning**

## **K - Means Clustering**

```python
# Annual Income vs Spending Score
# Create dataset
data = {
 'Annual_Income': [15, 18, 20, 22, 25, 60, 65, 70, 75, 80],
 'Spending_Score': [80, 75, 85, 70, 78, 30, 35, 25, 40, 20]
}
# Create DataFrame
df = pd.DataFrame(data)
print(df)
# Select input features
X = df[['Annual_Income', 'Spending_Score']]
# Create K-Means model with 3 clusters
model = KMeans(n_clusters=3, random_state=42, n_init=10)
# Train the model and assign clusters
df['Cluster'] = model.fit_predict(X)
# Display results
print("\nCustomer Clusters:")
print(df)
# Visualize clusters
plt.scatter(df['Annual_Income'], df['Spending_Score'],
 c=df['Cluster'], cmap='rainbow')
plt.xlabel('Annual Income')
plt.ylabel('Spending Score')
plt.title('Customer Segmentation using K-Means')
plt.show()
```
